# Position Matters: Statistical Profiles Across the PLL - Data Pre-Processing

This is all the data cleaning and pre-processing done for the EDA of Premier Lacrosse League Data.

## Data Pre-processing
Data sourced from *Premier Lacrosse League* website's player stats for 2025 Regular Season (https://stats.premierlacrosseleague.com/player-table).

A significant amount of time is taken to pre-process the data. It is an important aspect of any data analysis.

### Import necessary libraries

```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
```

### Import datasets

```python
raw_data = pd.read_csv("pll-player-stats.csv")
raw_data
```

```python
#copy contents of data into new variable to manipulate it
df = raw_data.copy()
df
```

### Check for null data

```python
#check data types of the imported file to ensure accurate and complete import
df.info()
```

There are 251 rows of data for each of 45 columns of data. This means there are 251 players in the Premier Lacrosse League in 2025. There are only 5 missing values in the `Jersey` field, so the data is mostly complete.

```python
#investigate the missing data values
df[df["Jersey"].isna()]
```

```python
df.loc[[200,222,233,234,239],["First Name","Last Name", "Position", "Games Played"]]
```

Note that these players did not play in any games. This means they were on the roster but never played. They do not provide any insight into player statistics because they have no statistics. For the purposes of this analysis, we will *exclude* them.

```python
df_clean = df[df["Games Played"]>0].copy()
```

### Check positional statistics

Now, investigate the statistics that are positional. We need to determine whether Face-off and Goalie statistics are genuinely zero for non-specialists, or if the zeroes stand in for "Not Applicable."

```python
df_clean.groupby("Position")[["Faceoffs","Shots", "Saves"]].mean()
```

Observe that these statistics are legitimate. Attackmen (`A`) rarely play in faceoffs or as a goalie, so their statistics as a zero are realistic. Defensemen (`D`) sometimes play faceoffs, which explains the nonzero statistic there. Goalies (`G`) can sometimes shoot, but their role primarily consists of saving shots. 

### Check for duplicates and team name datatypes

```python
#Check if there are duplicated rows
print("Duplicated rows:", df_clean.duplicated().sum())

#Check if team names are consistent
print("Unique Team Names:", df_clean["Team"].nunique())
```

There are no duplicated rows, which is good. No action needed.

There are only eight teams in the PLL, so we investigate why there are nine values present.

```python
df_clean["Team"].unique()
```

There is an extra `Team` called `Player Pool`. In the PLL, the Player Pool consists of all active roster players, free agents, and recent college draftees. These are players who do not have a contract with a specific team at the time the statistics were collected. We investigate to see if there are players who have actually played games during the season in it.

```python
df_clean[df_clean["Team"]=="Player Pool"][["First Name", "Last Name", "Team", "Games Played"]]
```

The three players in the Player Pool category have played in a game each. Hence, we will include them in the Player Analysis. However, during the Team Analysis, the `Player Pool` team will be excluded because it is not a full team roster.


### Sanity check percentage columns

```python
print(df_clean.columns.tolist())
```

```python
#name the columns that are supposed to be percentage values
pct_columns = ["Shooting %", "Shots On Goal %", "2Pt Shots %", "2Pt Shots On Goal %", "Faceoff %", "Save %", "Clean Save %"]
percent_columns = df_clean[pct_columns]

for col in percent_columns:
    print(col, df_clean[col].min(), df_clean[col].max())
```

These values stay between 0 and 1, which is the requirement for percentages. No action needed.

### Last few sanity checks

```python
#Sanity check total goals is 1 point + 2 point goals

print("Total Goals Check:", (df_clean["Total Goals"] != (df_clean["1Pt Goals"] + df_clean["2Pt Goals"])).sum())

#Sanity check points is number of 1 pt goals + 2*number of 2 pt goals + assists
print("Points Check:", (df_clean["Points"] != (df_clean["1Pt Goals"] + 2*df_clean["2Pt Goals"] + df_clean["Assists"])).sum())
```

We ensure the calculations for total goals and total points are the expected functions for interpretation. They are, so no action is needed.

### Export new cleaned dataset to csv
Now the data is cleaned and prepared for our analysis, we can export the modified file to use.

```python
df_clean.to_csv("pll-player-stats-cleaned.csv", index = False)
```
